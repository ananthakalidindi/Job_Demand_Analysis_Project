```python
import pandas as pd
```

```python
df = pd.read_csv("04_Data_Validation_Cleaning/04_clean_data.csv")
print("Rows used for analysis:", len(df))
```

```python
summary = pd.DataFrame({
    "Metric": ["Total Records", "Total Columns", "Average Salary", "Average Demand Score"],
    "Value": [len(df), len(df.columns), round(df["Salary_INR_Annual"].mean(), 2), round(df["Demand_Score"].mean(), 2)]
})

print(summary.to_string(index=False))
summary.to_csv("06_Data_Analysis/06_analyze_data.csv", index=False)
```

```python
salary_by_job = df.groupby("Job_Title")["Salary_INR_Annual"].mean().reset_index()
salary_by_job.columns = ["Job_Title", "Average_Salary"]
salary_by_job = salary_by_job.sort_values("Average_Salary", ascending=False)

print("Top 10 jobs by average salary:")
print(salary_by_job.head(10).to_string(index=False))
salary_by_job.to_csv("06_Data_Analysis/salary_by_job.csv", index=False)
```

```python
experience = df["Experience_Level"].value_counts().reset_index()
experience.columns = ["Experience_Level", "Job_Count"]

print("Jobs by experience level:")
print(experience.to_string(index=False))
experience.to_csv("06_Data_Analysis/experience_demand.csv", index=False)
```

```python
descriptive = df[["Salary_INR_Annual", "Demand_Score"]].describe()
print("Descriptive statistics:")
print(descriptive)
descriptive.to_csv("06_Data_Analysis/descriptive_statistics.csv")
```

```python
correlation = df[["Salary_INR_Annual", "Demand_Score", "Posting_Year"]].corr()
print("Correlation:")
print(correlation)
correlation.to_csv("06_Data_Analysis/correlation_analysis.csv")
```
