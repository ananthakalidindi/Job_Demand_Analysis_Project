# Job Demand Analysis

## Stage 2 – Data Acquisition & Filtering

In this stage, we select job postings with a Demand Score of 70 or more.

```python
import pandas as pd
```

```python
df = pd.read_csv("01_Data_Loading_Reading/01_load_data.csv")

print("Rows before filtering:", len(df))
```

```python
print("Demand Score values before filtering:")
print(df["Demand_Score"].head())
```

```python
filtered = df[df["Demand_Score"] >= 70]

print("Rows after filtering:", len(filtered))
print("Demand score condition: 70 or more")
```

```python
print("First 5 filtered records:")
print(filtered.head().to_string(index=False))
```

```python
filtered.to_csv("02_Data_Acquisition_Filtering/02_filter_data.csv", index=False)
print("Stage 2 completed")
print("Saved file: 02_Data_Acquisition_Filtering/02_filter_data.csv")
```
