```python
import os

# The notebook is inside its stage folder. Move to the project folder if needed.
if os.path.exists("data/job_demand_analysis.csv"):
    pass
elif os.path.exists("../data/job_demand_analysis.csv"):
    os.chdir("..")
elif os.path.exists("/content/Job_Demand_Analysis/data/job_demand_analysis.csv"):
    os.chdir("/content/Job_Demand_Analysis")

print("Project folder:", os.getcwd())
```

```python
import pandas as pd
```

```python
raw = pd.read_csv("01_Data_Loading_Reading/01_load_data.csv")
filtered = pd.read_csv("02_Data_Acquisition_Filtering/02_filter_data.csv")
clean = pd.read_csv("04_Data_Validation_Cleaning/04_clean_data.csv")
skills = pd.read_csv("03_Data_Extraction/03_extract_skills.csv")

print("Raw rows:", len(raw))
print("Filtered rows:", len(filtered))
print("Cleaned rows:", len(clean))
```

```python
top_job = clean["Job_Title"].value_counts().index[0]
top_location = clean["Location"].value_counts().index[0]
top_industry = clean["Industry"].value_counts().index[0]
top_experience = clean["Experience_Level"].value_counts().index[0]
top_skill = skills.iloc[0]["Skill"]

print("Most common job role:", top_job)
print("Most common location:", top_location)
print("Most common industry:", top_industry)
print("Most common experience level:", top_experience)
print("Most common skill:", top_skill)
```

```python
results = pd.DataFrame({
    "Result": [
        "Raw dataset rows",
        "High demand job rows",
        "Cleaned dataset rows",
        "Most common job role",
        "Most common location",
        "Most common industry",
        "Most common experience level",
        "Most common skill",
        "Average annual salary",
        "Average demand score"
    ],
    "Value": [
        len(raw),
        len(filtered),
        len(clean),
        top_job,
        top_location,
        top_industry,
        top_experience,
        top_skill,
        round(clean["Salary_INR_Annual"].mean(), 2),
        round(clean["Demand_Score"].mean(), 2)
    ]
})

print(results.to_string(index=False))
results.to_csv("08_Results_Interpretation/08_results.csv", index=False)
```
