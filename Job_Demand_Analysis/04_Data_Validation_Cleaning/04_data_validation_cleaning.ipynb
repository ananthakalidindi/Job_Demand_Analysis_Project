## Stage 4 – Data Validation & Cleaning

In this stage, we check missing values and remove duplicate rows. Missing values are handled with simple Pandas commands.

```python
import pandas as pd
```

```python
df = pd.read_csv("01_Data_Loading_Reading/01_load_data.csv")

print("Rows before cleaning:", len(df))
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate rows before cleaning:", df.duplicated().sum())
```

```python
df["Education"] = df["Education"].fillna("Not Specified")
df["Salary_INR_Annual"] = df["Salary_INR_Annual"].fillna(df["Salary_INR_Annual"].mean())

clean = df.drop_duplicates()

print("Rows after cleaning:", len(clean))
print("Rows removed as duplicates:", len(df) - len(clean))
```

```python
print("Missing values after cleaning:")
print(clean.isnull().sum())
```

```python
clean.to_csv("04_Data_Validation_Cleaning/04_clean_data.csv", index=False)
print("Stage 4 completed")
print("Saved file: 04_Data_Validation_Cleaning/04_clean_data.csv")
```
