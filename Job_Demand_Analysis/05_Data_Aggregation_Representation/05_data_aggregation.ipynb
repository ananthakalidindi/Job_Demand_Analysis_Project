## Stage 5 – Data Aggregation & Representation

We group the cleaned data to see job demand by job title, location, and industry.

```python
import pandas as pd
```

```python
df = pd.read_csv("04_Data_Validation_Cleaning/04_clean_data.csv")
print("Cleaned rows used:", len(df))
```

```python
job_counts = df["Job_Title"].value_counts().reset_index()
job_counts.columns = ["Job_Title", "Job_Count"]

print("Top 10 job titles:")
print(job_counts.head(10).to_string(index=False))
```

```python
location_counts = df["Location"].value_counts().reset_index()
location_counts.columns = ["Location", "Job_Count"]

print("Top 10 locations:")
print(location_counts.head(10).to_string(index=False))
```

```python
industry_counts = df["Industry"].value_counts().reset_index()
industry_counts.columns = ["Industry", "Job_Count"]

print("Industry demand:")
print(industry_counts.to_string(index=False))
```

```python
aggregate = pd.concat([
    job_counts.assign(Category="Job Title").rename(columns={"Job_Title":"Item"}),
    location_counts.assign(Category="Location").rename(columns={"Location":"Item"}),
    industry_counts.assign(Category="Industry").rename(columns={"Industry":"Item"})
], ignore_index=True)

aggregate = aggregate[["Category", "Item", "Job_Count"]]
aggregate.to_csv("05_Data_Aggregation_Representation/05_aggregate_data.csv", index=False)
print("Stage 5 completed")
print("Saved file: 05_Data_Aggregation_Representation/05_aggregate_data.csv")
```
