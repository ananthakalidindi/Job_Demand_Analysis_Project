## Stage 3 – Data Extraction

We extract individual skills from the Skills column and count how many job postings contain each skill.

```python
import os

# The notebook is inside its stage folder. Move to the project folder if needed.
if os.path.exists("data/job_demand_analysis.csv"):
    pass
elif os.path.exists("../data/job_demand_analysis.csv"):
    os.chdir("..")
elif os.path.exists("/content/Job_Demand_Analysis/data/job_demand_analysis.csv"):
    os.chdir("/content/Job_Demand_Analysis")

print("Project folder:", os.getcwd())
```

```python
import pandas as pd
```

```python
df = pd.read_csv("01_Data_Loading_Reading/01_load_data.csv")
print("Rows read:", len(df))
```

```python
skill_list = []

for skills in df["Skills"]:
    if pd.notnull(skills):
        for skill in str(skills).split(","):
            skill_list.append(skill.strip())

print("Total skill entries:", len(skill_list))
```

```python
skill_counts = pd.Series(skill_list).value_counts()

skills = skill_counts.reset_index()
skills.columns = ["Skill", "Job_Count"]

print("Top 10 skills:")
print(skills.head(10).to_string(index=False))
```

```python
skills.to_csv("03_Data_Extraction/03_extract_skills.csv", index=False)
print("Total different skills:", len(skills))
print("Stage 3 completed")
```
