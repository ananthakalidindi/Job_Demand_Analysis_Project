```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("04_Data_Validation_Cleaning/04_clean_data.csv")
skills = pd.read_csv("03_Data_Extraction/03_extract_skills.csv")

print("Rows used:", len(df))
print("Skills available:", len(skills))
```

```python
data = df["Job_Title"].value_counts().head(10)
plt.figure(figsize=(10,5))
data.plot(kind="bar")
plt.title("Top 10 Job Roles")
plt.xlabel("Job Role")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("07_Data_Visualization/01_market_demand_radar.jpg")
plt.show()
print("Chart 1 created")
```

```python
data = skills.head(10)
plt.figure(figsize=(10,5))
plt.bar(data["Skill"], data["Job_Count"])
plt.title("Top 10 Skills")
plt.xlabel("Skill")
plt.ylabel("Job Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("07_Data_Visualization/02_top_skills_polar.jpg")
plt.show()
print("Chart 2 created")
```

```python
data = df.groupby("Posting_Year").size()
plt.figure(figsize=(8,5))
plt.plot(data.index, data.values, marker="o")
plt.title("Yearly Job Posting Trend")
plt.xlabel("Year")
plt.ylabel("Number of Jobs")
plt.grid()
plt.tight_layout()
plt.savefig("07_Data_Visualization/03_yearly_job_posting_trend.jpg")
plt.show()
print("Chart 3 created")
```

```python
data = df["Industry"].value_counts()
plt.figure(figsize=(10,5))
plt.bar(data.index, data.values)
plt.title("Job Demand by Industry")
plt.xlabel("Industry")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("07_Data_Visualization/04_job_demand_by_industry.jpg")
plt.show()
print("Chart 4 created")
```

```python
data = df["Location"].value_counts().head(10)
plt.figure(figsize=(10,5))
plt.barh(data.index, data.values)
plt.title("Top 10 Job Locations")
plt.xlabel("Number of Jobs")
plt.ylabel("Location")
plt.tight_layout()
plt.savefig("07_Data_Visualization/05_jobs_by_location.jpg")
plt.show()
print("Chart 5 created")
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Salary_INR_Annual"] / 100000, df["Demand_Score"], alpha=0.5)
plt.title("Salary vs Demand Score")
plt.xlabel("Salary (Lakhs per year)")
plt.ylabel("Demand Score")
plt.tight_layout()
plt.savefig("07_Data_Visualization/06_salary_demand_market_map.jpg")
plt.show()
print("Chart 6 created")
```

```python
correlation = df[["Salary_INR_Annual", "Demand_Score", "Posting_Year"]].corr()
plt.figure(figsize=(7,5))
plt.imshow(correlation, cmap="viridis")
plt.colorbar()
plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45, ha="right")
plt.yticks(range(len(correlation.columns)), correlation.columns)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.savefig("07_Data_Visualization/07_correlation_matrix.jpg")
plt.show()
print("Chart 7 created")
```

```python
data = df.groupby("Experience_Level")["Salary_INR_Annual"].mean() / 100000
plt.figure(figsize=(8,5))
plt.bar(data.index, data.values)
plt.title("Average Salary by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Average Salary (Lakhs)")
plt.tight_layout()
plt.savefig("07_Data_Visualization/08_average_salary_by_experience.jpg")
plt.show()
print("Chart 8 created")
```
